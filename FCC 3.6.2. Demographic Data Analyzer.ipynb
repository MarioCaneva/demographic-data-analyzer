#Reading database

```python
import pandas as pd

# Load the data with proper formatting
df = pd.read_csv('/content/drive/MyDrive/adult.data.csv', header=None, names=[
    "age", "workclass", "fnlwgt", "education", "education-num",
    "marital-status", "occupation", "relationship", "race", "sex",
    "capital-gain", "capital-loss", "hours-per-week", "native-country", "salary"
], skipinitialspace=True)

# Let's check the data types
print(df.dtypes)
print(df.head())
```

# Race Count

```python
# Count how many people of each race are represented in this dataset
race_count = df['race'].value_counts()

# Display the result
print("Number of each race:\n", race_count)
```

# Average age of men

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
print(df['age'].isna().sum(), "non-numeric entries found in 'age' column.")
```

```python
average_age_men = df[df['sex'] == 'Male']['age'].mean().round(1)
print("Average age of men:", average_age_men)
```

# People with bachelor's degree

```python
# Calculate the percentage of people who have a Bachelor's degree
percentage_bachelors = (df['education'] == 'Bachelors').mean() * 100
percentage_bachelors = round(percentage_bachelors, 1)

# Display the result
print(f"Percentage with Bachelor's degrees: {percentage_bachelors}%")
```

# People with advanced education who earn >50K

```python
# Define advanced education
higher_education = df[df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]

# Calculate percentage of people with advanced education who earn >50K
higher_education_rich = (higher_education['salary'] == '>50K').mean() * 100
higher_education_rich = round(higher_education_rich, 1)

# Display the result
print(f"Percentage with higher education that earn >50K: {higher_education_rich}%")
```

# People without advanced education who earn >50K

```python
# Define non-advanced education group
lower_education = df[~df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]

# Calculate percentage of people without advanced education who earn >50K
lower_education_rich = (lower_education['salary'] == '>50K').mean() * 100
lower_education_rich = round(lower_education_rich, 1)

# Display the result
print(f"Percentage without higher education that earn >50K: {lower_education_rich}%")
```

# Minimum of hours working per week

```python
# Find the minimum number of hours worked per week
min_work_hours = df['hours-per-week'].min()

# Display the result
print(f"Min work time: {min_work_hours} hours/week")
```

# Min work-time with max earn

```python
# Filter people who work the minimum number of hours
min_workers = df[df['hours-per-week'] == min_work_hours]

# Calculate the percentage of those earning >50K
rich_percentage = (min_workers['salary'] == '>50K').mean() * 100
rich_percentage = round(rich_percentage, 1)

# Display the result
print(f"Percentage of rich among those who work fewest hours: {rich_percentage}%")
```

# Country with highest percentage of rich:


```python
# Group by country and calculate the percentage of people earning >50K
country_earnings = df.groupby('native-country')['salary'].apply(lambda x: (x == '>50K').mean() * 100)

# Identify the country with the highest percentage
highest_earning_country = country_earnings.idxmax()
highest_earning_country_percentage = round(country_earnings.max(), 1)

# Display the results
print("Country with highest percentage of rich:", highest_earning_country)
print(f"Highest percentage of rich people in country: {highest_earning_country_percentage}%")
```

# Popular occupation for those who earn >50K in India

```python
# Filter for people from India earning >50K
rich_indians = df[(df['native-country'] == 'India') & (df['salary'] == '>50K')]

# Find the most frequent occupation among them
top_IN_occupation = rich_indians['occupation'].mode()[0]

# Display the result
print("Top occupation in India for those who earn >50K:", top_IN_occupation)
```

# Demographic Data Analyzer Calculator

```python
import pandas as pd

def calculate_demographic_data(print_data=True):
    # Read data from file
    df = pd.read_csv('/content/drive/MyDrive/adult.data.csv', header=None, names=[
        "age", "workclass", "fnlwgt", "education", "education-num",
        "marital-status", "occupation", "relationship", "race", "sex",
        "capital-gain", "capital-loss", "hours-per-week", "native-country", "salary"
    ])

    # Clean whitespace from object (string) columns
    df = df.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)

    # Force numeric columns to correct dtype, coercing errors
    for col in ["age", "education-num", "capital-gain", "capital-loss", "hours-per-week"]:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # Drop rows with NaNs in important columns
    df.dropna(subset=["age", "hours-per-week", "salary"], inplace=True)

    # How many of each race are represented in this dataset?
    race_count = df['race'].value_counts()

    # What is the average age of men?
    average_age_men = round(df[df['sex'] == 'Male']['age'].mean(), 1)

    # Percentage of people with a Bachelor's degree
    percentage_bachelors = round((df['education'] == 'Bachelors').mean() * 100, 1)

    # Advanced education
    higher_education = df[df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]
    lower_education = df[~df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]

    higher_education_rich = round(
        (higher_education['salary'] == '>50K').mean() * 100, 1
    )

    lower_education_rich = round(
        (lower_education['salary'] == '>50K').mean() * 100, 1
    )

    # Minimum hours per week
    min_work_hours = df['hours-per-week'].min()

    # Percentage of rich among people who work minimum hours
    min_workers = df[df['hours-per-week'] == min_work_hours]
    rich_percentage = round(
        (min_workers['salary'] == '>50K').mean() * 100, 1
    )

    # Country with highest % of rich
    country_counts = df['native-country'].value_counts()
    rich_by_country = df[df['salary'] == '>50K']['native-country'].value_counts()
    rich_percentage_country = (rich_by_country / country_counts * 100).dropna()
    highest_earning_country = rich_percentage_country.idxmax()
    highest_earning_country_percentage = round(rich_percentage_country.max(), 1)

    # Top occupation in India for >50K
    top_IN_occupation = df[
        (df['native-country'] == 'India') & (df['salary'] == '>50K')
    ]['occupation'].mode()[0]

    # DO NOT MODIFY BELOW THIS LINE
    if print_data:
        print("Number of each race:\n", race_count)
        print("Average age of men:", average_age_men)
        print(f"Percentage with Bachelors degrees: {percentage_bachelors}%")
        print(f"Percentage with higher education that earn >50K: {higher_education_rich}%")
        print(f"Percentage without higher education that earn >50K: {lower_education_rich}%")
        print(f"Min work time: {min_work_hours} hours/week")
        print(f"Percentage of rich among those who work fewest hours: {rich_percentage}%")
        print("Country with highest percentage of rich:", highest_earning_country)
        print(f"Highest percentage of rich people in country: {highest_earning_country_percentage}%")
        print("Top occupations in India:", top_IN_occupation)

    return {
        'race_count': race_count,
        'average_age_men': average_age_men,
        'percentage_bachelors': percentage_bachelors,
        'higher_education_rich': higher_education_rich,
        'lower_education_rich': lower_education_rich,
        'min_work_hours': min_work_hours,
        'rich_percentage': rich_percentage,
        'highest_earning_country': highest_earning_country,
        'highest_earning_country_percentage': highest_earning_country_percentage,
        'top_IN_occupation': top_IN_occupation
    }

# Run it
calculate_demographic_data()
```
